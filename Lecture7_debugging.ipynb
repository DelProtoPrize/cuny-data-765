In the past several weeks, you've learned how to write basic Python programs. In this process, you have made different mistakes and seen a lot of `Error` messages. This is the most common thing in programming. Even experienced programmers need to debug their code constantly. 

First of all, you need to recognize and understand the nature of different types of bugs:

#### Programming bugs can be roughly classified into four categories, each with a different level of difficulty:

|         | **Obvious**  | **Hidden** |
|:--------|:---------|-----------:|
|**Persistent** | Easy |  Medium to difficult |
|**Occassional**| Medium | Difficult! |

**Obvious v.s. Hidden:**
- Obvious: you will definitely know your code is buggy because it **raises error messages or runs forever**. 
- Hidden: no obvious sign of bugs, but the **returned outputs may be incorrect**.

**Persistent v.s. Occassional:**
- Persistent: bugs **occur everytime** when you run the code.
- Occassional: bugs **occur only under certain conditions** (e.g. for certain specific inputs).

## Obvious bugs <a id=1.2></a>

### 1.Obvious & persistent bugs are the easiest ones to deal with. Just read (and google, if you don't understand) the error messages! Below are a few common errors:

`SyntaxError`: Often due to carelessness， e.g. using illegal variable names, forgetting a colon sign, a parenthesis, a comma, quotation marks, correct indentation levels etc.

```python
for i in [1, 3, 4, 'foo' 7, 9]
    print(i)
```

`IndexError`: you attempt to access elements beyond the range of a sequence e.g. a string / list / tuple / range etc. 

```python
s = 'Python Fundamentals'      # len(s) is 18
s[19]
```

`TypeError`: the operations you specified do not work with the data type. 

```python
s = 'Python Fundamentals'
s[4] - s[0]      
```

**There are many other types of errors:**

`AttributeError` : you tried to use an non-existent attribute (built-in methods) of an object.<br>
`KeyError`: you tried to access a non-existent dictionary key.<br>
`UnboundLocalError`: you tried to refer to a local variable before defining it. (see previous section on variable scope.)<br>
`ZeroDivisionError`: you trided to devide a number by zero. 

Please also check out the list of [Python built-in exceptions](https://docs.python.org/3/library/exceptions.html) to get a sense of other types of errors. 

### 2.  Obvious & occasional bugs are still relatively easy to handle. You need to consider different inputs that may trigger error warnings, and accordingly, design *test cases* to check the functionality of your code. 

**Most importantly, we need to partition the inputs (test cases) into a few different sets based on the logic of the operation.** 

For instance, when you write a function to calculate the square root of a number, it is wise to test your code with 
- some random negative numbers, 
- 0, 
- 1, 
- some random positive numbers.  

These test cases are partioned based on the mathmatic logic of calculating square root. 

**This principle may be adjusted to different scenarios. For instance, in the case of loops,** you want to include at least three test cases:

1. a case that will not execute the loop at all;
2. a case that executes the loop exactly once;
3. a case that executes the loop multiple times;

**If it is a control flow with multiple conditional branches, then you need to test every possible path in the code with different test cases.**

Simply put, you need to be clear about the logic of your operation, and design and partition test cases that follow the logic in related to the functionality of your program.

<div class = "alert alert-warning">
    
<b>Remember the rules of partitioning logic. You are going to follow this principle to design test cases in future applications.</b>

#### Input Validity

In addition, when you make assumptions about the inputs of a function, **you may want to check the validity of inputs first, and convert them to the valid format if necessary.** 


For instance:

```python
def remove_letter(s, remove=['a', 'e', 'i', 'o', 'u']):
    '''
    INPUTS:
    s: a string of any length.
    remove: a list of alpha letters to remove. Defaults to 5 vowels (a, e, i, o, u).
    
    OUTPUTS:
    the substring in s after removing all letters in `remove`.
    '''
    
    if isinstance(s, str):
    
        sub = ''
    
        for i in s:
            if i.lower() not in remove:
           # sub += i
                sub = sub + i
    
        return sub
    
    else:
        return 'the input S has to be a string.'
```

```python
remove_letter(765)
```

## Hidden bugs: <a id=1.3></a>

For hidden bugs, they are usually highly dangerous and difficult to debug because you may not even realize the code was buggy until one day you find out that the outputs were not what you expected. 

### In search of hidden bugs, try the following:

0. Enhance your **google skills** to look for useful pointers. 
1. Don't ask why the code is not doing what you want. **Ask how the code returns this specific output.** 
2. The bug often lies somewhere you did not expect:
    - **Expose intermediate stages of code execution with `print()` statements.**
    - In each step of your program, **how is your input data really being processed?**
    - Narrow down your search based on the results.
3. Test, test, test, with **carefully designed test cases** (see above section on test cases).

### When testing and debugging, do not change code directly!
- Backup your code
- **Change and test on a copy**
- Compare different versions until everything work well.

### a review of `for` loops

```python
## simple for loop

l1 = [1, 2, 3]

for i in l1:        
    print(i)         
```

```python
## for loop with range(len()) setup

l1 = [1, 2, 3]

for i in range(len(l1)):        
    print(i)                # i points to the numeric index 
    print(l1[i])            # we can use i to slice the list!
```

```python
## two lists of same length, do a **pair-wise** operation

l1 = [1, 2, 3]
l2 = ['a', 'b', 'c']            # two lists of the SAME length

for i in range(len(l1)):        # i points to the numeric index of l1, but l1 and l2 share the same index
    print(l1[i], l2[i])         # so i can be used to slice both l1 and l2!   
```

```python
## two lists, do an **all-pairs** operation (nested for loops)
## lists may not be of the same length

for i in [1, 2, 3]:               # 3 loops
    for j in ['a', 'b', 'c']:     # 3 loops
        print(i, j)               # 3 x 3 = 9 loops
```

$$x_i^2 + y_j$$

for **every possible pair of $x_i$, $y_j$ in the two lists**. Then the function returns the outputs as a list. 

*Note: Do **NOT** assume that both lists have the same length.* 

For instance, if `x = [1, 2]` and `y = [1, 2, 3]`, then your function should return a list `[2, 3, 4, 5, 6, 7]`, which is 

$[x_1^2+y_1,  x_1^2+y_2, x_1^2+y_3, x_2^2+y_1, x_2^2+y_2, x_2^2+y_3]$

```python
# a buggy solution:

my_list = []

def list_oper(x,y)
    '''
    add docstring here
    '''
    for i in range(len(x)):
    for j in range(len(y)):
        my_list.append(i**2 + j)
```

### First, make a copy and work on the copy only:

```python
my_list = []

def list_oper(x,y) :
    '''
    add docstring here
    '''
    for i in range(len(x)):
        for j in range(len(y)):
            my_list.append(i**2 + j)
            print(my_list)
```

### Finally, design a full set of test cases with proper partitioning logic:

```python
# test cases:

print(list_oper([], []))                              # not excute loops at all
print(list_oper([3], []))                             # excute x loop 1 time and y loop 0 time
print(list_oper([], [4]))                             # excute x loop 0 time and y loop 1 time
print(list_oper([1], [2]))                            # excute both loops once
print(list_oper([1, 2], [1, 2, 3]))                   # excute both loops multiple times
```

**Final Solution:**

<p style='color:red'> This question tests 1) function definition, 2) nested for loops, and 3) consideration of all possible scenarios. Particularly, what values do you expect if one or both of the input lists are empty? 
<br>
<br>
You as a creator of the function have the power to decide what you would like to do. You may specify that your function does not allow empty lists as inputs. You may also try to compute in one way or another if inputs are empty. But you need to demonstrate that you have considered different scenarios to preempt unexpected errors.</p>

```python
# Solution 1

def list_oper(x, y):
    '''
    INPUT:
    x, y: lists of numbers. Lists can be of different length but **cannot be empty**.
    
    OUTPUT:
    ans: the list of values of x_i**2 plus y_j for every possible pair of x_i and y_j in the two lists.
    '''
    ans = []
    
    if len(x) == 0 or len(y) == 0:
        return 'Input lists cannot be empty.'
    
    else:
        for i in x:
            for j in y:
                ans.append(i**2 + j)
        return ans
```

```python
# test cases

print(list_oper([], []))                              # not excute loops at all
print(list_oper([3], []))                             # excute each loop once and 0 time each
print(list_oper([], [4]))
print(list_oper([1], [2]))                            # excute both loops once
print(list_oper([1, 2], [1, 2, 3]))                   # excute both loops multiple times
```

```python
# Solution 2

def list_oper(x, y):
    '''
    INPUT:
    x, y: lists of numbers. Lists can be of different length, including empty.
    
    OUTPUT:
    ans: the list of values of x_i**2 plus y_j for every possible pair of x_i and y_j in the two lists.
    if x is empty, the function returns list y. 
    if x is not empty but y is empty, the function returns a list of x_i**2.
    '''
    ans = []
    
    if len(x) == 0:
        return y
    
    elif len(y) == 0:
        for i in x:
            ans.append(i**2)
            print(ans)
        return ans
    
    else:
        for i in x:
            for j in y:
                ans.append(i**2 + j)
                print(ans)
        return ans
```

```python
# test cases

print(list_oper([], []))                              # not excute loops at all
print(list_oper([3], []))                             # excute each loop once and 0 time each
print(list_oper([], [4]))
print(list_oper([1], [2]))                            # excute both loops once
print(list_oper([1, 2], [1, 2, 3]))                   # excute both loops multiple times
```

## Coding Style <a id=4></a>

The last thing we need to cover is coding style. 

"Style" may sound pretentious, but a good coding style is not only for aesthetics. It also helps you to write better structured programs, keep your code readable and maintainable, and preempt annoying bugs!  

### First, and most importantly, break up your code by writing small pieces of functions.

This is the second time we mentioned this. The first time was during debugging. 

In programming, it is always a good idea to divide code into **modules**, and: 
- write small pieces of reusable, self-contained functions; 
- test each piece before you combine them into a complex of functions. 

This way, you do not need to suffer the frustration of debugging a few hundred lines of messy code all at once. You can isolate and eliminate bugs before you finish writing the entire program. Your code is also organized and coherent this way. 

### Second, and equally importantly,  comment your code for critical steps and important parameters, and provide docstrings for each function you write.

Otherwise, you may fail to decipher your own code just in a week or two. And you will be desperately asking yourself: 

What is this? Why did I do this last time? Why is it not working now?

### Third, follow the good practices of writing Python.

In Python, opinions about what is "good" code are surprisingly consistent, such that those good practices are summarized in a document called [**PEP8 (Python Enhancement Proposal)**](https://pep8.org/), a formal style guide for python code. 

The guide is long, but useful. Some highlights include:

- Use **4 whitespaces** per indentation level.
- Operators should be separated from variables **by one space**: `a + b = c` instead of `a+b=c`
- Inline comments should be separated by **at least two spaces** from the statement. They should start with a `#` and a single space.

There are many more. Try to follow them whenever possible.
