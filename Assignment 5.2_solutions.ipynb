# Q1: Sum up numbers. (1.5 pts)

Please write a function `selective_sum(l)` to sum up **every other numbers** in a list, no matter it is a list or a list of lists (nested container). Your function takes one argument `l` which is a list (or a list of lists) of numbers (`int` or `float`). It returns the sum of every other numbers starting at the first number. 

For example, if the actual list passed into the function is `l = [0.3, 4, 6, [8.5, 9,] 1.7, [5, 3.2, 9,] 6, 8]`, then your funtion should sum up `0.3, 6, 9, 5, 9, 8` and return `37.3` as the output. 

**Hints:**

1. You can safely assume that the list (or list of lists) contains only numbers, either int or float.

2. You could first create another list that contains all numbers from `l` in precisely the **same order** but **without any nest lists**. Then go from there.

3. In order to do what described in bullet point #2, you can use `for` loops combined with `if...else...` statements to retrieve individual elements from different levels of containers.

````python
# your code below:

def selective_sum(l):
    '''
    INPUT:
    l: a list of lists of numbers, int or float.
    
    OUTPUT:
    the sum of every other numbers in the flattened l.
    '''    
    ans=[]                              # initialize an empty list
    for i in l:                         # for each element in l
        if type(i) == list:             # if the item is a list. Equivalent to "if isintance(i, list)"
            for j in i:                 # use a for loop to obtain each element in this sub-list and append to ans
                ans.append(j)           
        else:                           # if the item is NOT a list, i.e. if the item is a number
            ans.append(i)               # directly append the number to ans

    return sum(ans[::2])                # sum -> built-in function; slice notation slices every other element in ans. 
````

````python
# please run this cell after you complete Q1:
print(selective_sum([]))
print(selective_sum([7]))
print(selective_sum([11.5, 6, [3], 2]))
print(selective_sum([0.3, 4, 6, [8.5, 9,], 1.7, [5, 3.2, 9,], 6, 8]))
````

# Q2: Count occurences. (1.5 pts)

Please write a function `freq(str, target, overlap=True)` to check the number of times a non-empty string `target` appears in another string `str` (it could be a string of any length, including empty). Though `target` and `str` may contain both upper and lower cases, the result should NOT be case sensitive. 

The function `freq()` takes 2 positional arguments: `str` is a string of any length, and `target` is an non-empty string. It also takes 1 keyword argument: `overlap`, a boolean that defaults to `True`.

If `overlap` is set to `True` (default), the function `freq` returns a number, which is the occurences of `target` in `str`, including overlapping cases. For example, if we make a function call `freq('abcSosoSosed', 'sos')`, then it should return `3` as output.

If `overlap` is set to `False` , the function `freq` returns the occurences of `target` in `str` without overlapping cases. For example, if we make a function call `freq('abcSosoSosed', 'sos', overlap=False)`, then it should return `2` as output.

**Hints**: 
1. You need a `if...else...` statement to deal with the two `overlap` situations separately (either `True` or `False`).

2. Note that the built-in string method `s.count()` only works for non-overlapping cases. For instance, `'sosos'.count('sos')` will return 1. But if we count overlapping cases, then `'sos'` appears twice in `sosos`.

3. To solve the problem of overlapping cases, you may want to use the `range(len())` setup. This way, you can rely on slice notation to split  `str` and compare its segments with `target`.

````python
# your code below

def freq(strg, target, overlap=True):
    '''
    INPUT:
    strg: string of any length. 
    target: string, cannot be empty. 
    overlap: whether to count overlapping cases, defaults to False. 
    
    OUTPUT:
    The function returns n, the number of times that target appears in str, \
    including overlapping cases when overlap is set to True.
    '''
    
    # standardize strings for comparison
    strg = strg.lower()
    target = target.lower()                         # change both inputs to lowercases 
    
    n = 0                                           # initialize the counter as 0
    if overlap == True:                             # equivalent to "if overlap:"
        for i in range(len(strg)):                        # i points to the numeric index of each element in strg  
            if strg[i:i+len(target)] == target:           # slices out N=len(target) elements from str. Slicing moves forward by 1 step in each loop, starting from 0.
                n += 1                                    # compare the sliced elements with target, if same, counter plus 1.   
    
    else:                                           # when overlap sets to False 
        n = strg.count(target)                      # use built-in method to count non-overlapping cases
           
    return n  
````

````python
# run this cell after you complete Q2:

print(freq('', 'a'))
print(freq('c', 'cs'))
print(freq('soc', 'o'))
print(freq('abcsosoSoSed', 'sos'))
print(freq('abcsosoSoSed', 'sos', overlap=False))
print(freq('renqwkhEndenfdaboendfe', 'EnD'))
print(freq('renqwkhEndenfdaboendfe', 'EnD', overlap=False))
````
