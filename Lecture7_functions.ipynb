# Variable Scope <a id=1></a> 

Let's conclude functions with one last important point: the scope of variables.

Recall that a **variable assignment** connects a name to an object using the assignment operator `=`. After the assignment, the object can be referenced by the name. This is called a variable. 

**A "scope" is an enviroment (also called *namespace*) where variable names are assigned to objects and are thus recognizable and accessible.** 

There are four major types of variable scope, from the smallest to the largest:

- Local 
- Enclosing 
- Global 
- Built-in.

<img src = "https://res.cloudinary.com/dyd911kmh/image/upload/f_auto,q_auto:best/v1588956604/Scope_fbrzcw.png" align = center width = 300>

<p style="color: red"> General rule: variables defined in smaller scopes *cannot* be used in larger scopes, but variables defined in larger scopes are still *valid* in smaller scopes.</p>

A **built-in scope** is the largest scope. It contains the names of all of Python’s built-in objects (green / green bold keywords). This scope and all associated objects are created by the Python interpreter when it starts up. You do not need to define them but they are available at all times from anywhere as long as Python is running.

An **enclosing scope** appears in nested functions (inner functions) where you define a function inside other functions. 

We will be focusing on the **global** and **local** scope of variables:

## Global v.s. local scope <a id=1.1></a> 

- **Global scope**: 

The global scope is created when the **main program** (think of this jupyter notebook) starts. It contains any names that **you define** in the main program. Upon variable assignment, the names are recognizable and accessible anytime until the objects become orphaned, or until the main program terminates. 

**Before we learned functions, most code we saw were running in the global scope.** 

- **Local scope**: 

A local scope is created whenever **a function executes**. Any **variables created inside** the function are **only valid within the function.** 

The scope is local to the function, and **only exists until the function terminates**. Just like orphaned objects, Python would reclaim the memory allocated for the local scope and all name references would become invalid afterwards. 

```python
p1income = 1000                            # global scope

def household_income():                    # local scope
    p2income = 3000    
    hincome = p1income + p2income          # global variable is accessible/usable in local scope
    return hincome                 
```

```python
household_income()
```

```python
print(p1income)                        # global var, still accessible
```

```python
print(p2income)                        # local var, destroyed once the function ends
```

<div class = "alert alert-warning">
<b>Important takeaway</b> 
<br>
<br>
Global variables are still valid and accessible in a local scope, but local variables are <b>not</b> accessible in the global scope, and will be destroyed once the function ends.
</div>

## `UnboundLocalError` <a id=1.2></a>

The existence of multiple scopes means that **a certain variable name, like `x`, can exist in different scopes**. How does Python know which `x` to use? 

When you **call** the variable `x` in a function (inside a local scope), Python searches for `x` **from the inside out**, strictly in the order below:

**L**ocal scope: Python first searches for `x` in the local scope to see if it is defined inside the function.<br>
**G**lobal scope: If the variable `x` cannot be found in the local scope, Python looks for it in the global scope.<br>

*Note that we skipped **E**nclosing scope and **B**uilt-in scope for simplicity. Check out the [**LEGB rule**](https://realpython.com/python-scope-legb-rule/#:~:text=The%20LEGB%20rule%20is%20a,%2C%20and%20built%2Din%20scope) to learn more.*

An `NameError` will be raised if `x` cannot be found anywhere. 

**Under no circumstances would a variable created in a smaller scope be accessed and used in a larger scope** (unless you used the Python `global` or `unlocal` keywords, which are NOT recommended).

### Then what if you really try to modify a global variable within a local scope?

Let's take a look at a function that perhaps looks similar to other funcs we just saw:

```python
hincome = 0                                  # global scope

def household_income2(pincome):
    hincome = hincome + pincome              # where to look for the definition of "hincome"?
    return x
```

```python
household_income2(1000)
```

This `UnboundLocalError` looks funky. But you've seen a somewhat similar error in the global scope when you call a variable before you define it.

```python
a = a + 4                   # NameError
```

### So how to understand the `UnboundLocalError`?

When you call a variable in a local scope (i.e. a function), Python will first search for its definition in the local scope. Only when the variable is **NOT** defined in the local scope would Python reach out to larger scopes on the outside. 

In our example, we call `hincome` inside the function `household_income2()` when the line `hincome = hincome + pincome` executes. **Python starts from the right side of the `=` operator.** It sees an `hincome`. So it searches for the definition of `hincome` inside the function. Then it sees that the variable happens to be defined in the same line!

Since `hincome` *is* defined within the local scope, Python would NOT go to the global scope to look for it. So, inside the function, the variable assignment tries to assign `hincome` to `hincome` without giving `hincome` a value first (recall the `NameError` in the global scope). 

That's why we encounter a `UnboundLocalError`.
