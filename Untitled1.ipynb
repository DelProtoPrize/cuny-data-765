```python
import random
```

```python
def simulate_dynasty_lottery_top8(teams):
    """
    Simulates a 14-team dynasty draft lottery where only the top 8 
    non-playoff teams are eligible for the lottery (picks 1-3), and the 
    bottom 6 teams are locked into their draft positions.

    Args:
        teams: A list of team names in order of their regular season finish 
               (worst to best).

    Returns:
        A list representing the draft order after the lottery.
    """

    # 1. Separate Lottery-Eligible and Non-Lottery Teams
    lottery_teams = teams[:8]  # Top 8 teams are eligible
    non_lottery_teams = teams[8:]  # Bottom 6 teams are not

    # 2. Define Lottery Odds (for the 8 eligible teams)
    lottery_odds = {
        0: 140,  # 8th worst team (14.0% equivalent)
        1: 140,  # 7th worst (14.0% equivalent)
        2: 140,  # 6th worst (14.0% equivalent)
        3: 125,  # 5th worst (12.5% equivalent)
        4: 105,  # 4th worst (10.5% equivalent)
        5: 90,   # 3rd worst (9.0% equivalent)
        6: 75,   # 2nd worst (7.5% equivalent)
        7: 60    # Worst team (6.0% equivalent)
    }

    # 3. Create Lottery Balls (only for eligible teams)
    lottery_balls = []
    for i, team in enumerate(lottery_teams):
        lottery_balls.extend([i] * lottery_odds[i])

    # 4. Conduct the Lottery for the Top 3 Picks
    drawn_teams = []
    for _ in range(3):
        draw = random.choice(lottery_balls)
        drawn_teams.append(lottery_teams[draw])

        # Remove all balls for the drawn team
        lottery_balls = [ball for ball in lottery_balls if ball != draw]

    # 5. Determine the Rest of the Lottery Order
    remaining_lottery_teams = [team for team in lottery_teams if team not in drawn_teams]

    # 6. Combine the Results
    draft_order = drawn_teams + remaining_lottery_teams + non_lottery_teams

    return draft_order

# Example Usage:
teams = [
    "LALongshafts", "Scirt", "elitebandit", "jibrail", "PrimeProv", "hegotgame90210",
    "shawn5000", "RavioliTime", "MalcDaddy", "SayKwon", "wayneprice", "SpikeSpiegel0",
    "LordMalc25", "Dzel45"
]  # Replace with your actual team names, worst to best

draft_order = simulate_dynasty_lottery_top8(teams)
```

```python
print("Dynasty Draft Lottery Results (Top 8 Lottery):")
for i, team in enumerate(draft_order):
    print(f"{i+1}. {team}")
```
