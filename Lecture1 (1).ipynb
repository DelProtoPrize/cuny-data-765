## For each cell, either a code or markdown cell, there are two work modes: command mode, and edit mode. 

### Use edit mode to type:

```python
# when your mouse cursor are *inside* a cell, no matter a code cell or a markdown cell, you are in the "edit mode". 

# You can see the green border of the cell, as well as a small pencil on the top right corner of the notebook! 

10 + 11 

10 > 11

# texts
```

### Use command mode to manipulate the notebook:

By contrast, **when your mouse cursor are *outside* of the cells** (just click anywhere outside), notice that the border turns from green to blue, the pencil disappears, and you are now in the **command mode**!

In command mode, you can use a lot of convenient keyboard commands to work on the notebook: 

Press `a` to add a cell above. 

start typing...

Press `b` to add a cell below. 

Press `dd` to delete a cell. 

Press `x` to cut a cell, and press `v` to paste the cell!

Press `m` to turn the cell into a **markdown cell**, and press `y` to turn it back to a **code cell** !

#### Note! Those commands would only work when you are in command mode, i.e. cursor outside of the cell.  If you are *inside* a cell, the only command you can use is:

`shift` + `enter` 

to run the current cell (no matter markdown or code), show outputs (if any), and jump to the next cell below.

```python
# we will be heavily using code cells to practice python

## prepare packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

## set parameters
np.random.seed(765)
N = 1000

## generate random data
d = {'x': np.random.normal(1, 0.5, N), 'y': np.random.normal(1, 0.5, N)}
df = pd.DataFrame(data=d)

## inspect dataframe
df.head(20)
```

```python
## get descriptive statistics of the variables

df.describe()
```

```python
## draw a simple histogram to show the distribution of x

df.x.hist()
```
