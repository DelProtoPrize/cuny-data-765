# Structured Data Types: Dictionaries <a id=4></a>

In the last class, we saw sequence types like lists and strings, both of which can be indexed and sliced. Now we turn to **non-sequence structured types.** 

## Dictionaries

As the term `dict` indicates, a dictionary allows you to map pairs of items together ("a look-up table"). Therefore, each entry in a `dict` is consisted of two parts:

1. keys, or the names of elements: should be unique, immutable types, such as `str`, `int`, `float`
2. values of entries: can be duplicates, either mutable or immutable types

Dicts themselves are **mutable**, you may add or remove items inside dicts. 

```python
## Two common ways to define a dict:

d1 = {'food': [5, 3], 'service': ['good', 'bad']}     # key1:val1, key2:val2

d2 = dict(food=[5, 3], service=['good', 'bad'])       # key1=val1, key2=val2 

print(d1) 
print(d2)
print(type(d1))
```

```python
# add new entries to the dict

d1['location'] = ['Queens', 'Manhattan']                    
```

```python
d1
```

```python
d2 = {}                              # create an empty dict
type(d2)
```

#### Note that dicts are non-sequence types, meaning they are not indexed and do not support the use of slice notion.  For instance,  `dict[0]` returns an error. 
So how to access specific elements in dictionaries? We can call the associated keys using `dict[key]`

```python
d1[0]
```

```python
d1['food']
```

There is a safer way to access the values of an entry:

```python
d1.get('food')
```

Why do we need this safe mode? 

```python
d1['price']
```

```python
d1.get('price')
```

```python
d1.get('price', 'no price found')
```

**One thing to note:** 

Although a **dict** itself is a **non-sequence** type, some **values of the entries** in the dict can be **sequence types**, and they still support common Python index and slice notation!

```python
d1
```

```python
# again, multi-slicing for nested containers

d1['location'][0][1:3]
```

```python
## you can also modify the values of an entry if they are mutable types!
 
d1['food'][0] = 2                
d1['food']
```

Many familiar operations and methods are available for `dict`, such as `len(d)`, `in`, `del`. Check the [Python dictionary documentation](https://docs.python.org/3/library/stdtypes.html#mapping-types-dict) for more details. 

A few unique `dict` methods are worth attention:
- use `.keys()` to return a `dictview` object,  an iterable of all the keys
- use `.values()` to return a `dictview` object,  an iterable of all the values of the entries
- use `.items()` to return a `dictview` object, an iterable of all the items as (key, value) tuples

```python
d1
```

```python
d1.keys()
```

```python
d1.values()
```

```python
d1.items()
```

```python
'good' in d1        # check for membership ONLY IN the KEYS!
```

#### `dict`s are most likely nested containers. Therefore you should be careful about membership checking.  

```python
'good' in d1['service']
```

```python
'g' in d1['service'][0]
```

# More Structured Types <a id=5></a>

Lists and dictionaries are the most commonly used container types in Python. But there are also some other types that may come in handy at times. So let's briefly talk about two more types: tuples and sets. 

## Tuples <a id=5.1></a>

A `tuple` is similar to a list. The primary difference is that lists are **mutable** whereas tuples are **immutable**.

In other words, `tuple` can be seen as an "immutable" list where the elements can NOT be modified. It does most things that lists would do but is less flexible. So why do we still need tuples? 

In fact, tuples are immutable and *hashable*, and are thus more memory efficient and often faster than lists. You may also avoid accidental modification when programming.

```python
## common ways to declare a tuple

t = (1, 4, 'hello', [31.5, 74], 'a'<'b')           # use parenthesis

t = tuple([1, 4, 'hello', [31.5, 74], 'a'<'b'])    # or type conversion

type(t)
```

```python
()                       # empty tuple 
```

```python
1,                 # this is also a (less obvious) tupe
```

`tuple` is immutable -> the most distinguishing characteristics. 

```python
t[0]
```

```python
t[0] = 100
t
```

### Sets
`set` is similar to `list` and is also **mutable**. The difference is that sets are **not** sequenced/indexed, and they are **without duplicates**. 

```python
## two common ways to define a set:

s1 = {1, 1, 1, 2, 'a', 'a', 'a', 'z'}                        # what's the difference between the notations of sets and dicts?
s2 = set([1, 1, 1, 2, 'a', 'a', 'a', 'b'])                   # type conversion

print(type(s1), type(s2))
```

```python
s1             # no duplicates!
```

```python
s2
```

No duplicates is an important feature of `set`. This comes in handy when you need to **check the number of unique items in a container.**

You can use `set()` to convert another structure type to a `set`. Remember that this step removes all duplicates. Then you can check the length of the set and get the number of unique items. 

```python
l = [7, 7, 3, 6, 3, 2, 6, 8]

print(' original length: ', len(l), '\n', 'num of unique numbers: ', len(set(l)))
```

Some of the operations on lists and tuples also work for sets, such as `len()`, `in`, `not in`. Note that, however, sets are not a sequence type (not indexed) and any methods dealing with positions **cannot** be used on sets, including indexing and slicing. 

```python
s1[0]
```

#### For more methods on sets (and frozenset, which is both unordered AND immutable), check [Python set documentation](https://docs.python.org/3/library/stdtypes.html#set-types-set-frozenset)
