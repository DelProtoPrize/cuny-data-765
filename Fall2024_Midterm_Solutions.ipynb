### Q2. Consider the following piece of code: what is the output of ` m + n`? (Single Choice)

```Python
m = [1, 3, 5]
n = [2, 4, 6]
k = m
k[2] = 2
```

#### Correct Answer: D


A. `[3, 7, 11]`

B. `[3, 7, 8]`

C. `[1, 3, 5, 2, 4, 6]`

D. `[1, 3, 2, 2, 4, 6]`

E. None of the above.

#### Explanation:

This question tests your understanding of mutability and list operations:

Since `k = m` , both `k` and `m` are pointing to the same object.  When we modify `k[2]`, we also modify `m[2]` through aliasing (**mutability**). Therefore, `m` becomes `[1, 3, 2]` after we let `k[2] = 2`. 

`m + n` is thus `[1, 3, 2] + [2, 4, 6]`, and **the `+` operation** means **concatenation** for lists -- 

The answer is therefore `[1, 3, 2, 2, 4, 6]`.

### Q5. Consider the program below. After we run the code, what would be the values of `a`, `b`, and `func(1, 4)`, respectively? (Single Choice)

```python

def func(x, y):
    '''
    inputs: x, y. Single numbers, int or float.
    outputs: the answer of some mathematical operations on x and y.
    '''
	a = 7
	ans = x + y - a - b 
	return ans

a = 2
b = 3

func(1, 4)
```

#### Correct Answer : D

A. `2`, `3`, `0`

B. `7`, `3`, `-5`

C. `7`, `3`, `0`

D. `2`, `3`, `-5`

E. Error.

 #### Explanation:
This question tests your understanding of function calls and variable scope. 

First, `a = 2` and `b = 3` are defined in the **global scope**. Therefore, regardless of the specific code inside the function, the values of `a` and `b` will **not** be changed by the function call. They will still be `2` and `3` respectively. Choices B and C are ruled out. 

Now let's take a look at the function call `func(1, 4)`. Plugging `1` as `x` and `4` as `y` to the function, we see that `ans = 1 + 4 - a - b`. `a` is defined as `7` in **the local scope**, so the function plugs in `7` for `a`. `b` is not found anywhere in the local scope, so Python reaches out to the global scope to look for `b`. It sees that `b` is defined as `3` in the global scope. Together:

`ans = 1 + 4 - 7 - 3` 

So the function call returns `-5` as output. Choice D is correct.

## Q 1.2 (2 pts)

Now, consider another string `s = "Anna? Anna, Anna!"`. Write a program to check if this string reads the same backward as forward. Save the output to a variable `ans`. Note that the result should NOT be influenced by punctuations, whitespaces, and/or case sensitivity.

**Hints**
1. This question is built on top of Q1.1. To check whether the string reads the same backward as forward, **you can compare the string and its reversed string.**
2. You should deal with **non-alphabet** charaters (such as punctuations and whitespaces) and **case sensitivity** before the comparison, otherwise your results will be incorrect.
3. Think of a way to **find only alphabet letters** from the string. You may use a `for` loop to check each element of the string, and keep only alphabets.
4. You have learned about how to check whether an element is an alphabet -- use either a string method or certain string operations.

````python
## your code below:

s = "Anna? Anna, Anna!"

s = s.lower()                # standardize input to lower cases first

s_clean = ''                 # initialize an empty string
for i in s:
    if i.isalpha():          # you could also use if i >= 'a' to find alphabets
        s_clean += i         # addition operations means concatenation for strings
        
s_reversed = s_clean[::-1]        # reverse cleaned string

ans = s_clean == s_reversed       # compare the cleaned string to the reversed string, save to ans
        
````

````python
## run this cell after you complete Q1.2

print(ans)
````

### Explanation:

The key to solving this problem is to differentiate alphabets from other characters. ## Q 1.3 (5 pts)

Finally, based on what you have learned from Q1.1 and 1.2, please write a function `is_reverse(s)`. The function takes one positional argument `s` as input, where `s` is a **string** of any length. Then it checks whether the string reads the same backward as forward. If it reads the same, the function will return `True`. Otherwise, the function will return `False`. 

Note that this string could be a word, a phrase, or a sentence. But the result should **NOT** be influenced by punctuations, whitespaces, and/or case sensitivity. 

For instance, if the input is `'Eve'`, then calling `is_reverse('Eve')` will return a `True`. If the input is `"No lemon, no melon."`, the funtion should also return a `True`. 

*Note: if the input is an empty string, you could simply return a `True`. Or you could deal with this situation separately at will.*

**Hints**
1. This question is built on top of both Q 1.1 and Q 1.2. So the hints for previous questions all apply here.
2. Don't forget about **input validity**.
3. Don't forget about **docstrings**.  -- **You will received reduced marks if you don't follow class examples to write proper docstrings!**

````python
# your code below:

def is_reverse(s):
    '''
    INPUT:
    s: string of any length, can be a word, phrase or sentence. 
    
    OUTPUT:
    True/False: whether the word/phrase/sentence reads the same backward as forward.
    Note: the function only checks the alphabets and all non-alphabets will be ignored.
    '''
    
    # check input validity, string only
    if not isinstance(s, str):
        return 'Invalid input: input has to be of string type.'   # you may raise an error as well

    
    # below is the same as Q1.2
    s = s.lower()
        
    s_clean = ''                 
    for i in s:
        if i.isalpha():          
            s_clean += i           
    
    return s_clean == s_clean[::-1]
````

````python
# run this cell after you solve Q 1.3. 

print(is_reverse(''))
print(is_reverse('a'))
print(is_reverse('Eve'))
print(is_reverse("Madam, I'm Adam."))
print(is_reverse('Python Fundamentals'))
print(is_reverse(666))
````
