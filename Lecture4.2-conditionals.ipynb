# Control Flow Overview <a id=1></a>

In Python, a control flow is typically consisted of:
1. **conditional programs (branching)**, and/or 
2. **looping programs (iterations)**.

Most python programs can be decomposed to a number of conditionals and loops applying to various data objects. When you grasp control flows in Python, you can already use Python to accomplish many complicated tasks. 

Let's begin with conditionals:

A **conditional program**, or branching program, is essentially a decision tree. It could be as simple as the banner picture of this jupyter notebook. But it could also look as complicated as this:

No matter how complicated it looks, it is still a conditional program that is based on a bunch of conditional tests that evaluate to `True` or `False`. 

# Conditionals: the `if...else...` statement <a id=2></a>
A conditional program is a workflow that your code follows to decide whether it execute certain statements or not. The decision is solely based on the outcome(s) of one or more conditional tests. 

In Python, a conditional program is usually implemented with the `if`, `elif`, and `else` keywords. The general syntax is:

```python
if <conditional test 1>: # Boolean: True or False
    ## True block
    ...
    ...
    
elif <conditional test 2>: # optional
    ## True block
    ...
    ...
    
elif <conditional test 3>: # optional
    ## True block
    ...
    ...
    
## add as many elif clauses as you need
    ...
    ...
else: # optional
    # False block
    ...
    ...
    
```

## The `if` Statements <a id=2.1></a>

The core of a `if` statement is the `if` keyword followed by a conditional test which always evaluates to a boolean output: either `True` or `False`. However, there are two other elements that are equally important: 

```python
if <conditional test>:       # Boolean: True or False
    ...
    ...                      # an indented block of code
```

### Colon `:` and indentation `(four spaces*)`

First, the syntax of a `if` statement (and actually all control flows in Python) is featured by a `:` following the conditional test, and an indented block of code right below the `:`. 

Indentation means there are whitespaces before each statement in the block. Note that they should be indented at the same level, i.e. have the same number of spaces (**usually four**).

All statements **below a `:` and indented at the same level** constitute a specific `if` statement block. If there is no more indentations, i.e. if we are at the same level as the `if` keyword, we are already outside of the `if` statement. 
 
**There is no explicit signs of the start and the end of a control flow (or the blocks inside of the flow) other than `:` and spaces!**

*Note: it is recommended to use four spaces or a tab for indentation, even though you can use other number of spaces.*

### The logic of a `if` statement:
If and only if the conditional test returns a `True`, the program will run the block of expressions indented right below the `if` statement. Otherwise it will skip the block, i.e., do nothing, and move on to other things outside of the `if` statement.  

<img src='https://files.realpython.com/media/t.78f3bacaa261.png' align=center width=400>

````python
if 1+1<2:                                # if + Boolean + colon
    print('I am wrong.')                 # spaces + statement:  take this action if True, otherwise skip.
````

````python
a = 1
b = 2

if a > b:                                     # boolean: true or false                   
    print('a is larger than b')
    print('I am inside the if block')         # indentations signal the start and the end of a "if" block
print('I am out of the if block!') 
````

## The `else` Clauses <a id=2.2></a>

**Optionally**, we can add a `else` block. If the conditional test in the `if` statement outputs `False`, the program will skip the indented block below `if` and move on to run the `else` block. Therefore, **the `else` keyword is NOT followed by any conditional test.**

```python
if <conditional test>: # Boolean: True or False
    ## True block
    ...
    ...
else: # optional
    # False block
    ...
    ...  
```

Together, the `if...else...` statement takes care of all possibilities of the conditional test (because a `boolean` test is either `True` or `False`!)

Remember that the `else` block is optional, and it is always at the end of the `if...else...` flow. Only if you need to do something when the condition evaluates to `False` should the `else` clause be included. 

````python
if 1+1<2:                                
    print('I am wrong.')                 
else:
    print('I am right')
````

````python
a = 1
b = 2

if a > b:
    print('a is larger than b!')               # only run the if block if True
else:
    print('a is not larger than b!')           # only run the else block if False  
print('I am out of the control flow!')              
````

## The `elif` Clauses <a id=2.3></a>

It's likely that one conditional test cannot exhaust all conditions that we want to consider. So, let's introduce one final piece of the `if` statement: it is also **optional**, and it is used when you want to test more than one conditions. That is the `elif` clause (short for "else if".)
```python
if <conditional test 1>: # Boolean: True or False
    ## True block
    ...
    ...
    
## add as many elif clauses as you need   
elif <conditional test 2>: # optional
    ## True block
    ...
    ...
    
else: # optional
    # False block
    ...
    ...
    
```

You can add **as many `elif` clauses as you want**.

````python
a = 1
b = 1

if a > b:
    print('a is larger!')
    
elif a == b:                               # as many elif block as you need       
    print('a is equal to b!')

else:                                      # at most one else block, at the very end
    print('a is smaller than b!')
        
    
print('I am out of the control flow!')     # indentations decide the start and the end of a control flow / each block
````

<div class="alert alert-block alert-warning">
    
<b>Caution!</b> <br>
<br>
<b>Ordering matters</b> in `if..elif..else` statements.<br> 
<br>
The execution of the entire program stops immediately after it meets the <b>first</b> `True` returned by either the `if` statement or any of the `elif` clauses. The rest of the conditional tests will NOT be tested, and of course NONE of the remaining statements will be executed. <br>
<br>
We call this a short-circuit evaluation. At most one of the blocks will be executed. It is the first <code>True</code> block (and any nested blocks inside it) that the program runs into, OR the <code>False</code> block below the <code>else</code> statement (if included) when all of the above conditional tests evaluate to <code>False</code>.<br>
<br>
If there is no <code>else</code> block and none of the conditional tests return a <code>True</code>, nothing / no blocks will be executed. We move on to other things outside of the control flow.
    
</div>

````python
x = 5

if x > 10:
    print('large')
    
elif x == 5:                          # ordering matters
    print('precisely five!')
    
elif 3 < x < 9:                       # as many elif as you'd like
    print('medium')
    
else:                                 # else is always at the end
    print('small')
    
print('outside of the flow')
````
