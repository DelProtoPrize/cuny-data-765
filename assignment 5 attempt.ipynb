Please write a function `selective_sum(l)` to sum up **every other numbers** in a list, no matter it is a list or a list of lists (nested container). Your function takes one argument `l` which is a list (or a list of lists) of numbers (`int` or `float`). It returns the sum of every other numbers starting at the first number. 

For example, if the actual list passed into the function is `l = [0.3, 4, 6, [8.5, 9,] 1.7, [5, 3.2, 9,] 6, 8]`, then your funtion should sum up `0.3, 6, 9, 5, 9, 8` and return `37.3` as the output. 

**Hints:**

1. You can safely assume that the list (or list of lists) contains only numbers, either int or float.

3. In order to do what described in bullet point #2, you can use `for` loops combined with `if...else...` statements to retrieve individual elements from different levels of containers.

```python
# your code below:


real_count = [0.3, 4, 6, [8.5, 2], 1.7, [5, 3.2, 9], 6, 8]

def list_unpack(var1):
    """
    Flattens a list that may contain other sublists.
    
    Arg:
        a(list): Containing either an int, a float, or a sublist.
        
        
        Returns:
            Singular or multiple int(s), or float(s) concatenated into a list. Can also return no value.
    """
    list_numbers = []
    for entry in var1:
        if type(entry) is list:
            list_numbers.extend(entry)
        else:
            list_numbers.append(entry)
    return list_numbers

def selective_sum(var1):
    """
    Adds every other entry together to return a sum.
    
    Args:
        a (list): The list of int(s) or float(s) to summate.
        
    Returns:
        int or float : The summation of every other element of the list, 
            starting with the entry at the 0 position.
    """
    input_list = list_unpack(var1)
    total = 0
    for i in range(len(input_list)):
        if i % 2 != 0:
            continue
        else:
            total += input_list[i]
    return total


selective_sum(real_count)


    
```
