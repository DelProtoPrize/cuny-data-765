Now, let's change the rule of grading. If the student get a streak of right answers, they will be rewarded with bonus points. Specifically, when the student get a <code>True</code>, the student will get 1 point. When they have a streak of two <code>True</code>s in a row, the student will get 1 + 2 which is 3 points. If they have a streak of N <code>True</code>s in a row, the student will get 1 + 2 + ... + N points. <br> 
<br>
However, if the student get a <code>False</code>, they get 0 point for the answer, and the streak breaks immediately. <br>
<br>

```python
## Your code below:  (you can add more cells if needed)
tp = 0 
streak_count = [0]
for x in ans :
    count = streak_count[-1]
    if x > 0 :
        count += 1
    else:
        count = 0
    streak_count.append(count)
    print(streak_count)
tp = sum(streak_count)
print(tp)
print(sum(ans))

isinstance?


        
        
                    
                    
        
   
            
        








```
